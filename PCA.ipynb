```python
import numpy as np
import pandas as pd

data = {
    "Feature1": [2.5, 0.5, 2.2, 1.9, 3.1, 2.3, 2.0, 1.0, 1.5, 1.1],
    "Feature2": [2.4, 0.7, 2.9, 2.2, 3.0, 2.7, 1.6, 1.1, 1.6, 0.9]
}

df = pd.DataFrame(data)
print("Original Data:\n", df)
```

```python
X = df.values

mean = np.mean(X, axis=0)
X_centered = X - mean

print("\nMean:\n", mean)
print("\nCentered Data:\n", X_centered)
```

```python
cov_matrix = np.cov(X_centered.T)

print("\nCovariance Matrix:\n", cov_matrix)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("\nEigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
```

```python
idx = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print("\nSorted Eigenvalues:\n", eigenvalues)
print("\nSorted Eigenvectors:\n", eigenvectors)
```

```python
k = 1
principal_components = eigenvectors[:, :k]

print("\nPrincipal Component:\n", principal_components)
```

```python
X_reduced = X_centered @ principal_components

print("\nReduced Data (1D representation):\n", X_reduced)
```

```python
explained_variance_ratio = eigenvalues / np.sum(eigenvalues)
k = 1
print("Information retained:", explained_variance_ratio[:k].sum() * 100)
```

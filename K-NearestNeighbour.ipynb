```python
import pandas as pd
df = pd.read_csv("D:/Learn_ML/ML_algos/datasets/Iris.csv")
df.head()
```

```python
y = df["Species"]
X = df.drop(columns=['Id' , 'Species'])
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.2)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train) 
X_test = scaler.transform(X_test)
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
y_train = y_train.to_numpy()
y_test = y_test.to_numpy()
print(y_train)
print(y_test)
```

```python

y_train.dtype
```

```python
y_train = y_train.reshape(-1 , 1)
y_test = y_test.reshape(-1 , 1)
```

```python
import numpy as np
```

```python
def euclidean_distance(x1 , x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))

def knn_predict(X_train, y_train, x_query, k):
    distances = []
    for i in range(len(X_train)):
        dist = euclidean_distance(X_train[i], x_query)
        distances.append((dist, y_train[i].item()))
    distances.sort(key = lambda x : x[0])
    top_k = distances[:k]
    labels = [label for _, label in top_k]
    prediction = max(set(labels), key=labels.count)
    return prediction
```

```python
k = np.sqrt(X_train.shape[0])
```

```python
results = []
for i in range(len(X_test)):
    pred = knn_predict(X_train, y_train, X_test[i], k=5)
    results.append(pred)
```
